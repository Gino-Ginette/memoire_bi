```python
# GRAPHIQUE DE DÉCOUPAGE CHRONOLOGIQUE TRAIN / TEST
# À insérer dans la section "Séparation Train/Test"

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
from datetime import datetime

# Configuration du graphique
fig, ax = plt.subplots(figsize=(14, 3))

# Définition des périodes
train_start = datetime(2021, 1, 1)
train_end = datetime(2025, 5, 1)      # Date de coupure (80%)
test_start = datetime(2025, 6, 1)     # Début du test
test_end = datetime(2025, 12, 1)      # Fin du test

# Conversion pour matplotlib
train_start_num = train_start.toordinal()
train_end_num = train_end.toordinal()
test_start_num = test_start.toordinal()
test_end_num = test_end.toordinal()

# Zone d'entraînement (80% des données)
ax.axvspan(train_start_num, train_end_num, alpha=0.3, color='#0D4F8B', label='Entraînement (80%)')

# Zone de test (20% des données)
ax.axvspan(test_start_num, test_end_num, alpha=0.3, color='#E67E22', label='Test (20%)')

# Ligne verticale de séparation
ax.axvline(x=train_end_num, color='red', linestyle='--', linewidth=2, label='Frontière train/test (mai 2025)')

# Ajout des étiquettes de période
ax.text(train_start_num + (train_end_num - train_start_num)/2, 0.5, 
        'Période d\'entraînement\n2021 → mai 2025', 
        ha='center', va='center', fontsize=11, color='#0D4F8B', fontweight='bold')

ax.text(test_start_num + (test_end_num - test_start_num)/2, 0.5, 
        'Période de test\njuin → décembre 2025', 
        ha='center', va='center', fontsize=11, color='#E67E22', fontweight='bold')

# Mise en forme
ax.set_xlim(train_start_num - 100, test_end_num + 100)
ax.set_ylim(-1, 1)
ax.set_yticks([])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_position(('outward', 10))

# Formatage de l'axe des X avec les années
ax.xaxis.set_major_locator(plt.MaxNLocator(6))
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: datetime.fromordinal(int(x)).strftime('%Y')))

ax.set_xlabel('Année', fontsize=12, labelpad=10)
ax.set_title('Découpage chronologique des données (80% entraînement, 20% test)', fontsize=14, fontweight='bold', pad=20)

# Légende
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.25), ncol=3, fontsize=10)

plt.tight_layout()
plt.savefig('decoupage_temporel_train_test.png', dpi=150, bbox_inches='tight')
plt.show()
```
