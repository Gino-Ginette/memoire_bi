# 1. Introduction
#### Modélisation avec Random Forest

**Objectif**

Cette étape consiste à entraîner un modèle Random Forest afin de prédire les quantités futures de carburant commandées par les stations-service.

Le modèle utilisera :
- les variables temporelles ;
- les comportements historiques ;
- les tendances de commande.

## 2. Importation des Bibliothèques

```python
# IMPORTATION DES BIBLIOTHEQUES

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

## 3. Chargement du dataset ML
Le dataset contient toutes les variables préparées et les features nécessaires au modèle.

```python
# CHARGEMENT DU DATASET

df_ml = pd.read_csv("../data/propre/dataset_ml.csv")

print(df_ml.shape)

display(df_ml.head())
```

```python
# Vérifier la distribution des quantités
print(df_ml['quantite_semaine'].describe().round(0))
```

# 4. Sélection des variables explicatives
Je définis ce que le modèle observe et ce qu’il doit prédire.

```python
#VARIABLES EXPLICATIVES — noms corrigés
features = [
    "StationEncoded",       # ancien : StationEncoded       ✓ inchangé
    "ProduitEncoded",       # ancien : ProduitEncoded        ✓ inchangé
    "mois",                 # ancien : Mois
    "semaine_annee",        # ancien : Jour / JourSemaine    → numéro semaine ISO
    "est_fin_mois",         # ancien : Weekend               → indicateur fin mois
    "Lag_1",                # ancien : Lag_1                 ✓ inchangé
    "Lag_7",                # ancien : Lag_7                 ✓ inchangé
    "Rolling_4sem",         # ancien : Rolling_7j            → rolling 4 semaines
    "Rolling_12sem",        # nouveau : tendance 3 mois
    "Rolling_Std_4sem",     # ancien : Rolling_Std_7j        → std 4 semaines
    "nb_commandes",         # nouveau : nb jours commandés dans la semaine
]

target = "quantite_semaine"  # ancien : "Quantité facturée"

X = df_ml[features]
y = df_ml[target]

print(f"X : {X.shape}")   # (nb_lignes, 11 features)
print(f"y : {y.shape}")   # (nb_lignes,)
print()
print("Aperçu X :")
display(X.head())
```

**Explication:** On choisit quelles colonnes seront utilisées par le modèle pour faire ses prédictions. On garde les variables temporelles (mois, semaine, fin de mois), les indicateurs historiques (commandes semaine précédente, 7 semaines avant, moyennes sur 4 et 12 semaines, écart-type), le nombre de commandes, et les encodages station/produit. **La variable à prédire est quantite_semaine.**   

Les variables explicatives retenues incluent des indicateurs temporels (mois, semaine de l’année, fin de mois), des caractéristiques décalées (Lag 1, Lag 7), des moyennes mobiles sur 4 et 12 semaines, l’écart-type glissant, le nombre de commandes par semaine, ainsi que les encodages des stations et des produits.  

Elle capture à la fois la saisonnalité courte (Lag 1), la saisonnalité hebdomadaire (Lag 7), les tendances à moyen terme (rolling 4 semaines) et les tendances longues (rolling 12 semaines). L'écart-type glissant permet d'intégrer la variabilité, ce qui est utile pour les comportements irréguliers.

## 5. Séparation Train/Test
Mon problème étant temporel, alorsDonc il faut :entraîner sur le passé ettester sur le futur.
Sinon : fuite de données et faux résultats.

```python
# SEPARATION TRAIN / TEST

split_index = int(len(df_ml) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Train :", X_train.shape)
print("Test :", X_test.shape)
```

**Explication :** On découpe les données en deux parties : 80% pour entraîner le modèle (données passées) et 20% pour tester ses performances (données futures). Cette séparation est faite dans l'ordre chronologique, car c'est une série temporelle.  

# 6. Construction du modèle Random Forest
Explication des paramètres

| Paramètre    | Rôle                         |
| ------------ | ---------------------------- |
| n_estimators | nombre d’arbres              |
| max_depth    | profondeur maximale          |
| random_state | reproductibilité             |
| n_jobs=-1    | utilise tous les processeurs |


```python
# CONSTRUCTION DU MODELE

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

print(rf_model)
```

**Explication:**   
Le modèle Random Forest est paramétré avec 100 arbres (n_estimators=100), une profondeur maximale de 10 (max_depth=10) afin de limiter le risque de surapprentissage, et une initialisation fixe (random_state=42) pour garantir la reproductibilité des résultats."

## 7. Entraînement du modèle

Le modèle apprend :les habitudes, les tendances et les relations historiques.  L’apprentissage consiste à ajuster les 100 arbres de décision sur l’ensemble d’entraînement, permettant au modèle de capturer les dépendances entre les variables explicatives et la variable cible.

```python
# ENTRAINEMENT

rf_model.fit(X_train, y_train)

print("Modèle entraîné avec succès.")
```

## 8. Prédictions
Le modèle tente ici d'estimer les futures commandes

```python
# PREDICTIONS

predictions = rf_model.predict(X_test)

print(predictions[:10])
```

## 9. Évaluation des performances

### 9.1 MAE
Le MAE mesure l’erreur moyenne absolue.

```python
# MAE

mae = mean_absolute_error(
    y_test,
    predictions
)

print("MAE :", mae)
```

**Explication :**
Le **MAE (Erreur Absolue Moyenne)** mesure en moyenne de combien le modèle se trompe en valeur absolue. Ici, **3399 litres sensiblement égal à 3 400L** d'erreur moyenne. Ce qui signifie qu’en moyenne, la prédiction s’écarte de la commande réelle de cette quantité."

### 9.2 RMSE

Pénalise fortement les grosses erreurs.

```python
# RMSE

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predictions
    )
)

print("RMSE :", rmse)
```

Le **RMSE (Racine de l'Erreur Quadratique Moyenne)** pénalise davantage les grosses erreurs. À 6292 litres, il est supérieur au MAE, ce qui indique la présence d'erreurs importantes.
L'écart MAE/RMSE confirme que le modèle échoue sur certaines commandes extrêmes (peut-être des pics de consommation ou des commandes exceptionnelles).

### 9.3 R² SCORE

Mesure la qualité globale du modèle.

```python
# R2 SCORE

r2 = r2_score(
    y_test,
    predictions
)

print("R² :", r2)
```

**Le coefficient de détermination R²** indique la proportion de la variabilité des commandes expliquée par le modèle. **0,67 signifie que 67%** de la variabilité est expliquée — le modèle est correct mais perfectible.

### 9.4 MAPE 

```python
# Calculer le MAPE pour contextualiser l'erreur
mape = (abs(y_test - predictions) / y_test).mean() * 100
print(f"Quantité moyenne  : {y_test.mean():.0f} litres")
print(f"MAE               : 3 400 litres")
print(f"MAPE              : {mape:.1f} %")
```

Le **MAPE (Erreur Absolue Moyenne en Pourcentage)** donne l'erreur relative moyenne : environ 43%. La commande moyenne étant de 12 098 litres, cela semble élevé, mais attention aux petites commandes.
Ce qui veut dire que sur commandes moyennes **(12 000 L), 43% = 5 000 L d'erreur**, ce qui est significatif 

## 10. Analyse des variables importantes
Cette étape montre quelles variables influencent le plus les commandes

```python
# IMPORTANCE DES VARIABLES

importance = pd.DataFrame({
    "Variable": features,
    "Importance": rf_model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

display(importance)
```

L’analyse d’importance des variables révèle que les **moyennes mobiles sur 4 et 12 semaines (Rolling_4sem et Rolling_12sem)** sont les principaux **déterminants des prédictions**, indiquant que le modèle s’appuie prioritairement sur les tendances historiques récentes et à moyen terme.

Elle est très utile pour la compréhension et la confiance dans le modèle. Le fait que les lags directs (Lag_1, Lag_7) aient moins de poids que les rolling suggère que la tendance lissée est plus informative que la simple observation ponctuelle.

```python
# VISUALISATION

plt.figure(figsize=(10,6))

plt.barh(
    importance["Variable"],
    importance["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Variables")

plt.title("Importance des variables")

plt.show()
```

## 11. Comparaison Réel vs Prédit

```python
# REEL VS PREDIT

comparaison = pd.DataFrame({
    "Réel": y_test.values,
    "Prédit": predictions
})

display(comparaison.head(20))
```

```python
# GRAPHIQUE rREEL VS PREDIT

plt.figure(figsize=(12,6))

plt.plot(
    y_test.values[:100],
    label="Réel"
)

plt.plot(
    predictions[:100],
    label="Prédit"
)

plt.legend()

plt.title("Réel vs Prédit")

plt.show()
```

La comparaison graphique des valeurs réelles et prédites confirme la tendance générale du modèle, tout en mettant en évidence certains pics sous-estimés ou surestimés, caractéristiques des limites des approches basées sur l’historique seul.  
On voit bien que le modèle peine sur les valeurs extrêmes (ex: ligne 16: 19000 réel vs 16221 prédit).

```python
# Distribution des erreurs

errors = y_test - predictions

plt.figure(figsize=(10,5))

plt.hist(errors, bins=30)

plt.title("Distribution des erreurs")

plt.xlabel("Erreur")
plt.ylabel("Fréquence")

plt.show()
```

### Commentaire et interprétation du graphique de distribution des erreurs

**Commentaire du graphique :**

La distribution des erreurs présente les caractéristiques suivantes :

Centre autour de zéro : La fréquence la plus élevée se situe autour de la valeur 0 sur l'axe des erreurs, ce qui indique que la majorité des prédictions sont proches des valeurs réelles.

Étendue des erreurs : Les erreurs s'étendent approximativement de -20 000 à +120 000, avec une queue plus longue du côté positif.

Asymétrie visible : La distribution n'est pas parfaitement symétrique : on observe une queue plus étendue vers les erreurs positives (jusqu'à 120 000) que vers les erreurs négatives (jusqu'à -20 000 seulement).

**Interprétation :**

Biais potentiel (sous-prédiction fréquente) :
L'asymétrie vers les erreurs positives suggère que le modèle a tendance à sous-estimer les valeurs réelles plus souvent qu'à les surestimer. Autrement dit, lorsque le modèle se trompe, il prédit des valeurs inférieures à la réalité, d'où une erreur positive (réel - prédit > 0).

Absence d'erreurs négatives extrêmes :
L'absence d'erreurs très négatives (inférieures à -20 000) indique que le modèle ne surestime jamais massivement la réalité. C'est un point rassurant pour la gestion des stocks : une surestimation massive pourrait conduire à un surstockage coûteux.

Erreurs positives exceptionnelles :
La présence d'erreurs allant jusqu'à 120 000, bien que peu fréquentes (fréquence faible sur l'axe Y), montre qu'il existe des cas où le modèle sous-estime très fortement la demande. Ces cas méritent une analyse approfondie (saisonnalité exceptionnelle, promotion, rupture chez un concurrent, etc.).

Concentration des erreurs :
La forte concentration des erreurs autour de zéro (fréquence culminant à environ 800-1000) indique une bonne performance globale du modèle, la majorité des prédictions étant précises.


***La distribution des erreurs est centrée sur zéro, indiquant une bonne précision globale du modèle. On observe toutefois une asymétrie vers les erreurs positives, signe d'une tendance à sous-estimer la demande plutôt qu'à la surestimer. Les surestimations massives sont absentes (pas d'erreur négative extrême), ce qui est favorable d'un point de vue gestion des stocks. En revanche, quelques sous-estimations importantes (jusqu'à 120 000) existent et devront être analysées cas par cas.***



```python
residus = y_test - predictions

plt.figure(figsize=(10,5))

plt.plot(residus.values)

plt.title("Analyse des résidus")

plt.show()
```

**Commentaire du graphique :**

Centre autour de zéro : La majorité des résidus se situent entre −20 000 et +60 000, avec une concentration autour de 0.

Absence de tendance visible : Les résidus ne montrent pas d’évolution systématique à la hausse ou à la baisse sur l’ensemble des 3 500 observations.

Quelques valeurs extrêmes : On observe des pics résiduels atteignant ponctuellement +120 000 et −120 000, mais leur nombre semble limité.

Pas de structure en entonnoir : La dispersion des résidus reste relativement homogène sur toute la plage (pas de variance clairement croissante ou décroissante).

**Interprétation :**

Hypothèse de linéarité / bon ajustement :
L’absence de tendance dans les résidus indique que le modèle capture correctement la structure principale des données. Il n’y a pas d’erreur systématique liée au temps ou à l’ordre des observations.

Homoscédasticité acceptable :
La dispersion constante des résidus suggère que la variance des erreurs est stable. Cela valide une hypothèse importante pour de nombreux modèles statistiques (régression linéaire, SARIMA, etc.).

Points atypiques :
Les pics extrêmes (±120 000) correspondent à des observations que le modèle n’a pas bien prédites. Leur rareté (quelques points sur 3 500) est rassurante, mais ils méritent d’être analysés individuellement (ex. : demande exceptionnelle, anomalie de saisie, effet ponctuel non modélisé).

Absence d’autocorrélation évidente :
Aucune succession régulière de signes positifs ou négatifs n’est visible à l’œil nu. Cela plaide en faveur de résidus indépendants, condition favorable à la fiabilité des intervalles de prédiction.

**Synthèse :**

Les résidus sont centrés sur zéro et ne présentent ni tendance ni dispersion croissante. Leur distribution est stable dans le temps, ce qui valide les hypothèses d’homoscédasticité et d’absence de biais systématique. Quelques résidus extrêmes apparaissent ponctuellement, sans remettre en cause la robustesse globale du modèle. Ces points atypiques feront l’objet d’une vérification individuelle.
